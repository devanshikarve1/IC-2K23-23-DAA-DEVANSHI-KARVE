```python
import random
import time
import matplotlib.pyplot as plt
import heapq

# 1. Generate Random Input
def generate_data(n):
    return [random.randint(1, 100000) for _ in range(n)]


# 2. Merge Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


# 3. Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + equal + quick_sort(right)


# 4. Heap Sort
def heap_sort(arr):
    heap = arr.copy()
    heapq.heapify(heap)

    sorted_arr = []

    while heap:
        sorted_arr.append(heapq.heappop(heap))

    return sorted_arr


print("All Three Sorting Algorithms Defined Successfully!")
```

```python
n_values = [100, 500, 1000, 2000, 5000]

merge_times = []
quick_times = []
heap_times = []

for n in n_values:

    data = generate_data(n)
    expected = sorted(data)

    # Merge Sort
    start = time.perf_counter()
    merge_result = merge_sort(data.copy())
    merge_time = time.perf_counter() - start
    merge_times.append(merge_time)

    # Quick Sort
    start = time.perf_counter()
    quick_result = quick_sort(data.copy())
    quick_time = time.perf_counter() - start
    quick_times.append(quick_time)

    # Heap Sort
    start = time.perf_counter()
    heap_result = heap_sort(data.copy())
    heap_time = time.perf_counter() - start
    heap_times.append(heap_time)

    # Verify Sorted Output
    assert merge_result == expected
    assert quick_result == expected
    assert heap_result == expected

    print("\nInput Size:", n)
    print("Merge Sort Time:", round(merge_time, 6), "seconds")
    print("Quick Sort Time:", round(quick_time, 6), "seconds")
    print("Heap Sort Time:", round(heap_time, 6), "seconds")
    print("All Three Outputs Verified Successfully!")

print("\nExecution Completed Successfully!")
```

```python
print("SORTING PERFORMANCE COMPARISON")

print("\nN\tMerge Sort\tQuick Sort\tHeap Sort")

for i in range(len(n_values)):
    print(
        n_values[i], "\t",
        round(merge_times[i], 6), "\t",
        round(quick_times[i], 6), "\t",
        round(heap_times[i], 6)
    )
```

```python
plt.figure(figsize=(10, 6))

plt.plot(n_values, merge_times, marker='o', label='Merge Sort')
plt.plot(n_values, quick_times, marker='o', label='Quick Sort')
plt.plot(n_values, heap_times, marker='o', label='Heap Sort')

plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Sorting Algorithm Performance Comparison")

plt.legend()
plt.grid(True)

plt.savefig("execution_time_graph.png")
plt.show()
```
